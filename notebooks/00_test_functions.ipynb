```python
import sys
sys.path.append('/home/pulpo/Documents/quantitative-trading')
from src.quanttrading.analysis import calculate_stats, detect_outliers
from src.quanttrading.data import prepare_data
import pandas as pd
```

```python
# DOWNLOAD DATA
df_nvda = prepare_data('NVDA', '6mo', '4h')
```

```python
print('Primeras filas')
print(df_nvda.head())

print('\nRETURN STATISTICS\n')

stats = calculate_stats(df_nvda['Return'])
for key, value in stats.items():
    print(f'{key}: {value: .4f}')

print('\nDetecting outliers\n')

outliers = detect_outliers(df_nvda['Return'], thereshold=3)

print(f'Total de dias:     {len(df_nvda)}')
print(f'Dias con outliers: {outliers.sum()}')
print(f'Porcentaje:       {(outliers.sum() / len(df_nvda) * 100): .2f}%')

print('\nDias con outliers (mostrando Return y distancia de STD):\n')
outlier_days = df_nvda[outliers][['Return']].copy()
outlier_days['Distance_from_mean (STD)'] = ((df_nvda['Return'] - stats['mean']).abs() / stats['std'])[outliers]
print(outlier_days)
```
